# Part 1


```python
import numpy as np

data = np.array([
    [80, 90, 85],   #rows= student's score
    [70, 75, 78],   #columns = subjects
    [95, 88, 92],
    [60, 65, 70],
])

print(data)
```

```python
scalar = np.array(95)  #scalar= a single element (one number)

vector = np.array([80, 90, 85]) #vector= a full row( one student's scores) or a full column(all students' scores in one subjec)

matrix = data #matrix= a table of rows and columns

semester2 = np.array([
    [85, 92, 88],
    [74, 80, 79],
    [97, 90, 94],
    [66, 70, 72],
])

tensor = np.array([data, semester2]) #tensor= tables stacked on top of each other


print("scalar:", scalar.ndim, scalar.shape)
print("vector:", vector.ndim, vector.shape)
print("matrix:", matrix.ndim, matrix.shape)
print("tensor:", tensor.ndim, tensor.shape,'\n')


a = np.array([1, 2, 3])        # (n,) = 1D array: a flat list of n elements (not a row, not a column)
b = np.array([[1], [2], [3]])   # # (n,1) = 2D array: a column vector with n rows and 1 column

print(' the difference between (n,) and (n,1):')
print("a:",a,'shape =' ,a.shape, 'number of dim=',a.ndim,'\n')
print("b:",b,'shape =' ,b.shape,'number of dim=', b.ndim)
```

```python
print('Apply element-wise operations') #when or how to use element-wise ? see the examples below

bonus = np.array([
    [5, 0, 2],
    [5, 3, 2],
    [0, 2, 0],
    [8, 5, 5],
])

print("Addition (data + bonus):\n", data + bonus, '\n')          # add bonus marks
print("Subtraction (semester2 - data):\n", semester2 - data, '\n') # improvement between semesters

weights = np.array([0.5, 0.3, 0.2])                                # Math 50%, Science 30%, English 20%
print("Multiplication (data * weights):\n", data * weights, '\n')

print("Division (data / 100):\n", data / 100)                     # scores as 0-1
```

```python
print('Aggregations:')

print("Sum:", data.sum())     # total of all scores
print("Mean:", data.mean())   # average score
print("Min:", data.min())     # lowest score
print("Max:", data.max())     # highest score
print("Std:", data.std())     # how spread out the scores are from the mean
```

```python
v = np.array([2,6,1,3])
w = np.array([3,-1,5,8])
print("v : ",v)
print("w : ",w)

print("Vector addition :", v + w)
print("Scalar multiplication :", v * w)

dot_vw = v @ w  # if > 0 They point to the same dircetion
                # if 0 They are perpendicular
                # if < 0 They point to opposite directions
print("Dot product :",dot_vw)

v_norm = np.linalg.norm(v)
w_norm = np.linalg.norm(w)
print("Norm v :",v_norm)
print("Norm w :",w_norm)

cosine_similarity = dot_vw / (v_norm * w_norm)
print("Cosine similarity :",cosine_similarity)
```

```python
#Create two vectors and test: Are they linearly independent? Do they form a basis for R2?
p = np.array([1, 2])
q = np.array([2, 4])   # dependent with p (q = 2 * p)
r = np.array([0, 1])   # independent direction

A_dep = np.column_stack([p, q])
A_ind = np.column_stack([p, r])

rank_dep = np.linalg.matrix_rank(A_dep)
rank_ind = np.linalg.matrix_rank(A_ind)

print("Rank([p q]) =", rank_dep)
print("Rank([p r]) =", rank_ind)

print("\n[p, q]: Linearly independent?", rank_dep == 2, ". Basis for R2?", rank_dep == 2)
print("[p, r]: Linearly independent?", rank_ind == 2, ". Basis for R2?", rank_ind == 2)
```

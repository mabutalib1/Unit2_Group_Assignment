# Mission 1 – Discovering Eigenvectors
Consider the matrix

|2 1|

|1 2|

Suppose the following vectors are given:

v1 = [1,1]
v2 = [1,-1]

```python
A = np.array([[2,1],[1,2]])

v1 = np.array([1,1])
v2 = np.array([1,-1])

Av1 = A @ v1
Av2 = A @ v2

print("A @ v1 =",A @ v1) # = 3 * v1 , same direction but streched by a factor of 3
print("A @ v2 =",A @ v2) # = v1 , keeps it the same

# They’re called eigenvectors because the matrix only scales them without changing their direction.
```

# AI Insight
Why do most vectors change direction while only a few remain on the same line?

The matrix stretches some directions more than others.

A normal vector is a mix of those directions. Its parts get stretched by different amounts, so the vector tilts.

An eigenvector points exactly along one of those directions. It just gets longer or shorter, with nothing to tilt it.

# AI Insight
What does an eigenvalue represent during a matrix transformation?


An eigenvalue is the amount the matrix scales its eigenvector.

- λ > 1: the vector is stretched (λ = 3 here, so v₁ becomes 3 times longer).
- λ = 1: the length stays the same (v₂ here).
- 0 < λ < 1: the vector shrinks.
- λ < 0: the vector flips to point the opposite way, along the same line.
- λ = 0: the vector collapses to zero.

The eigenvector gives the direction, and the eigenvalue gives how much that direction gets stretched.

# AI Insight
Why does every eigenvalue have its own eigenvector?

Each eigenvalue is a stretch amount, and its eigenvector is the direction that gets stretched by exactly that amount. One direction can’t be stretched by two different amounts, so each eigenvalue gets its own eigenvector.

# Mission 4: Geometric Interpretation



| Eigenvalue | Interpretation |
|---|---|
| **λ = 3** | **Stretched.** The eigenvector keeps the same direction and becomes **3 times longer** (λ > 1). |
| **λ = 0.5** | **Compressed.** The eigenvector keeps the same direction and becomes **half its length** (0 < λ < 1). |
| **λ = −2** | **Reversed and stretched.** The eigenvector **flips to the opposite direction** (λ < 0) and becomes **2 times longer** (\|λ\| = 2). It is still on the same line. |


### AI Insight
**How do eigenvalues affect scaling and direction?**
- The **size** of λ controls the **scaling**: |λ| > 1 stretches, |λ| < 1 compresses, |λ| = 1 keeps the length.
- The **sign** of λ controls the **direction**: λ > 0 keeps the direction, λ < 0 reverses it.
- λ = 0 collapses the vector to zero (that direction is lost, and the matrix is singular).
- In all cases the eigenvector **never rotates off its line**; it only stretches, shrinks or flips.
